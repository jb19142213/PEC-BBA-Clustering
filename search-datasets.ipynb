```python
import numpy as np
import tensorflow as tf
from sklearn.datasets import fetch_openml
import torch
import torchvision
import torchvision.transforms as transforms
from torchvision.models import resnet50, ResNet50_Weights
from tqdm import tqdm
import warnings
warnings.filterwarnings("ignore")

# ---------------------------
# 1. MNIST
# ---------------------------
(x_train_m, _), (x_test_m, _) = tf.keras.datasets.mnist.load_data()
mnist_data = np.concatenate([x_train_m, x_test_m], axis=0)
mnist_data = mnist_data.reshape(-1, 28*28).astype(np.float32) / 255.0
print(f"MNIST: {mnist_data.shape}")

# ---------------------------
# 2. Fashion-MNIST
# ---------------------------
(x_train_f, _), (x_test_f, _) = tf.keras.datasets.fashion_mnist.load_data()
fashion_data = np.concatenate([x_train_f, x_test_f], axis=0)
fashion_data = fashion_data.reshape(-1, 28*28).astype(np.float32) / 255.0
print(f"Fashion-MNIST: {fashion_data.shape}")

# ---------------------------
# 3. USPS
# ---------------------------
usps = fetch_openml(name='USPS', version=2, as_frame=True, parser='auto')
usps_data = usps.data.values.astype(np.float32) / 255.0
print(f"USPS: {usps_data.shape}")

# ---------------------------
# 4. STL-10 with ResNet-50 features (2048-dim as in paper)
# ---------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Extracting STL-10 features using ResNet-50...")

# Transform: resize to 224x224, normalize with ImageNet stats
transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Load STL-10 train and test splits
stl_train = torchvision.datasets.STL10(root='./data', split='train', download=True, transform=transform)
stl_test = torchvision.datasets.STL10(root='./data', split='test', download=True, transform=transform)
stl_full = torch.utils.data.ConcatDataset([stl_train, stl_test])

# Load pretrained ResNet-50
weights = ResNet50_Weights.IMAGENET1K_V1
model = resnet50(weights=weights)
model = model.to(device)
model.eval()

# Feature extractor: remove final fc layer, keep avgpool (global average pooling)
class FeatureExtractor(torch.nn.Module):
    def __init__(self, model):
        super().__init__()
        self.features = torch.nn.Sequential(*list(model.children())[:-1])  # keep avgpool, remove fc
    def forward(self, x):
        x = self.features(x)  # shape: (N, 2048, 1, 1)
        return x.view(x.size(0), -1)  # flatten to (N, 2048)

extractor = FeatureExtractor(model).to(device)

# Extract features in batches
batch_size = 64
dataloader = torch.utils.data.DataLoader(stl_full, batch_size=batch_size, shuffle=False, num_workers=2)
all_features = []
with torch.no_grad():
    for images, _ in tqdm(dataloader, desc="Extracting STL-10 features"):
        images = images.to(device)
        feats = extractor(images)
        all_features.append(feats.cpu().numpy())
stl_features = np.concatenate(all_features, axis=0)
print(f"STL-10 (features): {stl_features.shape}")  # Expected: (13000, 2048)
```

```python
import os
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.datasets import fetch_rcv1, fetch_openml
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer
from sklearn.preprocessing import LabelEncoder
import warnings
from scipy.sparse import csr_matrix
warnings.filterwarnings('ignore')

# Helper: file path mapping
file_path_map = {}
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        file_path_map[filename] = os.path.join(dirname, filename)

print("=== Preprocessing Start (Missing values preserved as NaN) ===")

# ===============================================
# 1. REUTERS (Option B)
# ===============================================
print("\n--- REUTERS (Option B) ---")
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.reuters.load_data()
x = np.concatenate([x_train, x_test], axis=0)
y = np.concatenate([y_train, y_test], axis=0)

# Build bag-of-words sparse matrix (no missing)
vocab_size = 30982
n_samples = len(x)
rows, cols, data = [], [], []
for i, seq in enumerate(x):
    counts = {}
    for word_idx in seq:
        counts[word_idx] = counts.get(word_idx, 0) + 1
    for word_idx, cnt in counts.items():
        rows.append(i)
        cols.append(word_idx - 1)  # to 0-index
        data.append(cnt)
bow = csr_matrix((data, (rows, cols)), shape=(n_samples, vocab_size))

tfidf = TfidfTransformer()
tfidf_matrix = tfidf.fit_transform(bow)
term_freq = np.asarray(bow.sum(axis=0)).ravel()
top_indices = np.argsort(term_freq)[-2000:]
reuters_features = tfidf_matrix[:, top_indices].toarray()

# Label mapping to 4 root categories (approximate)
category_mapping = {
    3:0,4:0,5:0,6:0,7:0,8:0,9:0,10:0,11:0,12:0,13:0,14:0,15:0,16:0,17:0,18:0,19:0,
    20:1,21:1,22:1,23:1,24:1,25:1,26:1,27:1,28:1,29:1,
    30:2,31:2,32:2,33:2,34:2,35:2,36:2,37:2,38:2,
    39:3,40:3,41:3,42:3,43:3,44:3,45:3
}
mask = np.array([category_mapping.get(l, -1) for l in y]) != -1
reuters_features = reuters_features[mask]
reuters_labels = np.array([category_mapping.get(l) for l in y if category_mapping.get(l, -1) != -1])
print(f"Reuters: {reuters_features.shape}, labels {len(np.unique(reuters_labels))} classes")



# ===============================================
# 3. Russian Housing (preserve missing)
# ===============================================
print("\n--- Russian Housing ---")
train_file = file_path_map.get('train.csv')
df_train = pd.read_csv(train_file, low_memory=False)
cols = ['full_sq', 'life_sq', 'kitch_sq', 'price_doc']
df_housing = df_train[cols].copy()

# Keep missing as NaN; do NOT impute.
# Only transform price (divide by 1e5) – this does not introduce NaN.
df_housing['price_doc'] = df_housing['price_doc'] / 1e5

# Convert to float32, preserving NaN
housing_features = df_housing.values.astype(np.float32)
print(f"Russian Housing: {housing_features.shape}")
print(f"  Missing count per column: {np.isnan(housing_features).sum(axis=0)}")

# ===============================================
# 4. Horse Colic (preserve missing)
# ===============================================
print("\n--- Horse Colic ---")
horse = fetch_openml(name='colic', version=1, as_frame=True)
df_horse = horse.frame

# Select numeric columns (keep all, including those with missing)
numeric_cols = df_horse.select_dtypes(include=[np.number]).columns.tolist()
# We also need the label column 'surgery' but keep it separate
if 'surgery' in numeric_cols:
    numeric_cols.remove('surgery')
X_horse = df_horse[numeric_cols].copy()
y_horse = df_horse['surgery'].copy()   # will be preserved as object (categorical)

# Keep missing as NaN; no imputation.
horse_features = X_horse.values.astype(np.float32)

# Encode labels: we keep NaN for missing labels (they remain as NaN)
le = LabelEncoder()
# To preserve NaN, we fill missing with a placeholder, encode, then set back to NaN?
# Simpler: keep labels as original with NaN, no encoding.
# But we'll encode for consistency: we'll treat missing as -1 (since it's not a valid category)
y_horse_encoded = np.full(len(y_horse), -1, dtype=np.int32)
mask = y_horse.notna()
y_horse_encoded[mask] = le.fit_transform(y_horse[mask].astype(str))
# Now y_horse_encoded has -1 for missing, and 0,1 for valid.
horse_labels = y_horse_encoded
print(f"Horse Colic: {horse_features.shape}, labels unique: {np.unique(horse_labels)}")

# ===============================================
# 5. NHANES (preserve missing)
# ===============================================
print("\n--- NHANES ---")
tables = ['demographics.csv', 'dietary.csv', 'examination.csv', 'laboratory.csv', 'questionnaire.csv']
merged = None
for table in tables:
    path = file_path_map.get(table)
    if path:
        df = pd.read_csv(path, low_memory=False)
        # Keep all columns (including categorical) but we'll only use numeric for now
        # To keep missing, we'll keep as is.
        if merged is None:
            merged = df
        else:
            merged = pd.merge(merged, df, on='SEQN', how='inner')

if merged is not None:
    # Keep all numeric columns (including those with missing)
    numeric_cols = merged.select_dtypes(include=[np.number]).columns.tolist()
    # Remove SEQN if present (it's an identifier, not a feature)
    if 'SEQN' in numeric_cols:
        numeric_cols.remove('SEQN')
    nhanes_features = merged[numeric_cols].values.astype(np.float32)
    print(f"NHANES: {nhanes_features.shape}")
    print(f"  Missing count per column (first 10): {np.isnan(nhanes_features).sum(axis=0)[:10]}")
else:
    print("NHANES: No tables found")
    nhanes_features = None

# ===============================================
# 6. HHAR (use edit_Phones_accelerometer.csv, preserve missing)
# ===============================================
print("\n--- HHAR ---")
hhar_file = file_path_map.get('edit_Phones_accelerometer.csv')
if hhar_file:
    sample_rows = 100000  # adjust as needed
    df_hhar = pd.read_csv(hhar_file, nrows=sample_rows)
    # Select numeric feature columns (exclude identifiers and labels)
    exclude = ['sample_id', 'Arrival_Time', 'Creation_Time', 'User', 'Model', 'Device', 'gt',
               'Arrival_Time2', 'Creation_Time2']
    feature_cols = [c for c in df_hhar.columns if c not in exclude and df_hhar[c].dtype in [np.float64, np.int64]]
    # Also drop columns that are datetime derived (if any)
    feature_cols = [c for c in feature_cols if not any(x in c for x in ['Arrival_', 'Creation_', 'sample_id'])]
    X_hhar = df_hhar[feature_cols].copy()
    # Keep missing (should be none, but just in case)
    hhar_features = X_hhar.values.astype(np.float32)
    # Labels
    le_hhar = LabelEncoder()
    y_hhar = df_hhar['gt'].fillna('unknown').astype(str)
    hhar_labels = le_hhar.fit_transform(y_hhar)
    print(f"HHAR (downsampled): {hhar_features.shape}, labels {len(np.unique(hhar_labels))}")
else:
    print("HHAR file not found")
    hhar_features = None
    hhar_labels = None

# ===============================================
# Save preprocessed data (preserving NaNs)
# ===============================================
print("\n--- Saving Preprocessed Data ---")
os.makedirs('/kaggle/working/preprocessed', exist_ok=True)

np.savez('/kaggle/working/preprocessed/reuters.npz', X=reuters_features, y=reuters_labels)
np.savez('/kaggle/working/preprocessed/russian_housing.npz', X=housing_features)
np.savez('/kaggle/working/preprocessed/horse_colic.npz', X=horse_features, y=horse_labels)
if nhanes_features is not None:
    np.savez('/kaggle/working/preprocessed/nhanes.npz', X=nhanes_features)
if hhar_features is not None:
    np.savez('/kaggle/working/preprocessed/hhar.npz', X=hhar_features, y=hhar_labels)

print("Done. Preprocessed files saved to /kaggle/working/preprocessed/")
```

```python
import numpy as np
import os

# Parameters
alpha = 0.1   # missing rate
seed = 42
rng = np.random.default_rng(seed)

# List of datasets to simulate missingness
datasets = [
    ('mnist.npz', 'mnist_missing.npz'),
    ('fashion_mnist.npz', 'fashion_mnist_missing.npz'),
    ('usps.npz', 'usps_missing.npz'),
    ('stl10.npz', 'stl10_missing.npz'),
    ('reuters.npz', 'reuters_missing.npz'),
    ('rcv1.npz', 'rcv1_missing.npz'),
    ('hhar.npz', 'hhar_missing.npz')
]

# Load each dataset, apply masking, save new file
for input_name, output_name in datasets:
    input_path = f'/kaggle/working/preprocessed/{input_name}'
    if not os.path.exists(input_path):
        print(f"Warning: {input_path} not found, skipping.")
        continue
    
    # Load
    data = np.load(input_path)
    X = data['X']
    y = data.get('y', None)  # some datasets may have labels
    
    # Create mask: random entries set to NaN
    mask = rng.random(X.shape) < alpha
    X_masked = X.copy()
    X_masked[mask] = np.nan
    
    # Save
    out_path = f'/kaggle/working/preprocessed/{output_name}'
    if y is not None:
        np.savez(out_path, X=X_masked, y=y)
    else:
        np.savez(out_path, X=X_masked)
    print(f"Saved {out_path} with shape {X_masked.shape}, missing rate: {np.isnan(X_masked).mean():.4f}")

print("Done.")
```

```python
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from notmiwae_pytorch import NotMIWAE, Trainer
from notmiwae_pytorch.utils import set_seed, impute
from notmiwae_pytorch.models.base import Encoder
import contextlib
import sys
from io import StringIO
from IPython.display import FileLink

# ---------------------------
# 1. Define two-layer encoder (architecture as per NotMIWAE paper)
# ---------------------------
class TwoLayerEncoder(Encoder):
    def __init__(self, input_dim, hidden_dim, latent_dim):
        super().__init__(input_dim, hidden_dim, latent_dim)
        self.backbone = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, hidden_dim),  # Second hidden layer
            nn.Tanh()
        )
        self.mu = nn.Linear(hidden_dim, latent_dim)
        self.logvar = nn.Linear(hidden_dim, latent_dim)

# Patch the encoder in the library module
import notmiwae_pytorch.models.notmiwae as notmiwae_module
notmiwae_module.Encoder = TwoLayerEncoder

# ---------------------------
# 2. Load Reuters data
# ---------------------------
# Assuming files are in /kaggle/working/preprocessed/ as specified
data_missing = np.load('/kaggle/working/preprocessed/reuters_missing.npz')
X_missing = data_missing['X']          # (N, D) with NaN
data_orig = np.load('/kaggle/working/preprocessed/reuters.npz')
X_orig = data_orig['X']                # Complete (N, D)

# Dynamically determine input dimension (e.g., 2000)
input_dim = X_missing.shape[1]
print(f"Detected Reuters input dimension: {input_dim}")

mask = np.isfinite(X_missing).astype(np.float32)
X_filled = np.nan_to_num(X_missing, nan=0.0)

X_filled = torch.tensor(X_filled, dtype=torch.float32)
mask = torch.tensor(mask, dtype=torch.float32)
X_orig = torch.tensor(X_orig, dtype=torch.float32)

dataset = TensorDataset(X_filled, mask, X_orig)
# Batch size 16 or 32 is generally fine for Reuters
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)

# ---------------------------
# 3. Model Configuration
# ---------------------------
set_seed(42)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

model = NotMIWAE(
    input_dim=input_dim,
    latent_dim=50,                # Standard for Reuters in NotMIWAE experiments
    hidden_dim=128,               
    n_samples=20,                 # K = 20 importance samples
    missing_process='selfmasking',
    out_dist='gauss',             # Use 'gauss' for TF-IDF / continuous features
).to(device)

# ---------------------------
# 4. Trainer
# ---------------------------
trainer = Trainer(
    model,
    device=device,
    lr=1e-3,
    original_data_available=True,   # Tracks RMSE vs X_orig
    log_dir='./runs/reuters_notmiwae'
)

print("Starting training on Reuters...")
history = trainer.train(dataloader)
print("Training finished.")

# ---------------------------
# 5. Save model
# ---------------------------
torch.save(model.state_dict(), '/kaggle/working/reuters_notmiwae_model.pth')
print("Model saved to /kaggle/working/reuters_notmiwae_model.pth")

# ---------------------------
# 6. Impute missing values (silent)
# ---------------------------
impute_dataset = TensorDataset(X_filled, mask)
with contextlib.redirect_stdout(StringIO()):
    X_imputed = impute(
        model,
        impute_dataset,
        n_samples=1000,           # High sample count for final reconstruction
        batch_size=64,
        device=device,
        verbose=False
    )
print(f"Imputation complete. Shape: {X_imputed.shape}")

# ---------------------------
# 7. Save results and provide link
# ---------------------------
np.savez_compressed('/kaggle/working/reuters_imputed.npz', X=X_imputed)
print("Imputed Reuters data saved to /kaggle/working/reuters_imputed.npz")
FileLink('/kaggle/working/reuters_imputed.npz')
```
